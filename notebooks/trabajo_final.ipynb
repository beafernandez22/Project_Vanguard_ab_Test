## 4. ¿En qué paso se concentran las visitas no completadas en cada diseño? - comparar Test y Control

```python
# Crear una tabla con el último paso y el grupo A/B

from Functions import resumir_visitas

visitas_resumen = resumir_visitas(web_ab_clean, exp_ab)

pd.crosstab(
    visitas_resumen.loc[
        ~visitas_resumen["completada"], "ultimo_paso"
    ],
    visitas_resumen.loc[
        ~visitas_resumen["completada"], "Variation"
    ]
)
```

Observamos que Test tiene menos visitas no completadas en todos los pasos, especialmente en step_3.

Sin embargo, todavía no podemos afirmar que el nuevo diseño sea mejor únicamente con estos números, porque los grupos tienen distintos tamaños y pueden generar diferentes cantidades de visitas.

Para compararlos correctamente, necesitamos calcular porcentajes sobre el total de visitas de cada grupo.

```python
# Calcular las tasas de visitas no completadas

# Total de visitas por grupo
total_visitas = visitas_resumen.groupby("Variation").size()

# Visitas sin confirmación por grupo
visitas_no_completadas = (
    visitas_resumen[~visitas_resumen["completada"]]
    .groupby("Variation")
    .size()
)

# Calcular porcentaje
tasa_no_completadas = (
    visitas_no_completadas / total_visitas * 100
).round(2)

print(tasa_no_completadas)
```

Los resultados sugieren que el rediseño de Vanguard facilita que los usuarios lleguen al final del proceso. Sin embargo, será necesario realizar pruebas estadísticas para comprobar si esta diferencia es significativa.

## 5. ¿Los clientes con más antigüedad tardan menos en completar el proceso?

```python
# Preparar los eventos de cada visita
eventos = web_ab_clean.sort_values(
    ["client_id", "visit_id", "date_time"]
)

# Primera fecha de inicio por visita
inicios = (
    eventos[eventos["process_step"] == "start"]
    .groupby(["client_id", "visit_id"])["date_time"]
    .min()
)

# Primera confirmación por visita
confirmaciones = (
    eventos[eventos["process_step"] == "confirm"]
    .groupby(["client_id", "visit_id"])["date_time"]
    .min()
)

# Unir ambas fechas
duraciones = pd.concat(
    [inicios.rename("inicio"), confirmaciones.rename("confirmacion")],
    axis=1
).dropna()

# Duración en minutos
duraciones["minutos"] = (
    duraciones["confirmacion"] - duraciones["inicio"]
).dt.total_seconds() / 60

# Conservar duraciones válidas
duraciones = duraciones[duraciones["minutos"] >= 0]

duraciones["minutos"].describe().round(2)
```

- La mitad de las visitas completadas tardan 4,23 minutos o menos.
- La media es superior a la mediana, lo que indica que algunas visitas muy largas elevan el promedio.
- Hay valores extremos, como una visita de aproximadamente 5 horas, que podrían corresponder a sesiones interrumpidas.
Por eso, para comparar clientes, conviene utilizar inicialmente la mediana, que es menos sensible a los valores extremos.

Ahora vamos a relacionar las duraciones con los años de antigüedad de cada cliente.

```python
# Incorporar los datos demográficos a las duraciones
duraciones_clientes = duraciones.reset_index().merge(
    clientes_ab[["client_id", "clnt_tenure_yr"]],
    on="client_id",
    how="left"
)

# Crear grupos de antigüedad
duraciones_clientes["grupo_antiguedad"] = pd.cut(
    duraciones_clientes["clnt_tenure_yr"],
    bins=[0, 5, 10, 15, 20, 100],
    labels=["0-4 años", "5-9 años", "10-14 años",
            "15-19 años", "20 o más"],
    right=False
)

# Calcular la mediana de duración por antigüedad
duraciones_clientes.groupby(
    "grupo_antiguedad", observed=True
)["minutos"].median().round(2)
```

### ¿Los clientes con más antigüedad tardan menos en completar el proceso?

Los resultados muestran una tendencia contraria a la hipótesis inicial: los clientes con mayor antigüedad presentan tiempos de finalización ligeramente superiores.

La mediana aumenta progresivamente desde 3,90 minutos para los clientes con menos de 5 años de antigüedad hasta 4,53 minutos para aquellos con 20 años o más.

Esto sugiere que una mayor experiencia como cliente de Vanguard no implica necesariamente completar el proceso online más rápidamente.

No obstante, sería necesario analizar otros factores, como la edad y el diseño utilizado, para comprender mejor estas diferencias.

```python
# Calcular la mediana de tiempo por antigüedad
tiempos_antiguedad = (
    duraciones_clientes
    .groupby("grupo_antiguedad", observed=True)["minutos"]
    .median()
)

# Crear gráfico
plt.figure(figsize=(9, 5))

sns.barplot(
    x=tiempos_antiguedad.index,
    y=tiempos_antiguedad.values,
    color="steelblue"
)

plt.title("Tiempo de finalización según antigüedad")
plt.xlabel("Antigüedad del cliente")
plt.ylabel("Tiempo mediano (minutos)")
plt.ylim(0, 6)

for i, valor in enumerate(tiempos_antiguedad.values):
    plt.text(i, valor + 0.1, f"{valor:.2f} min", ha="center")

plt.tight_layout()
plt.show()
```

Se aprecia perfectamente que el tiempo mediano de finalización aumenta progresivamente con la antigüedad del cliente: de 3,90 a 4,53 minutos.

```python
from statsmodels.stats.proportion import proportions_ztest
from Functions import calcular_tasa_errores, conclusion

# Obtener los datos actualizados del KPI 3
errores, tasa_errores = calcular_tasa_errores(web_ab)

# Número de visitas con retrocesos por grupo
exitos_err = errores.groupby("Variation")["error"].sum()

# Número total de visitas por grupo
totales_err = errores.groupby("Variation")["error"].count()

# Contraste de dos proporciones
z_stat, p = proportions_ztest(
    count=[
        exitos_err["Test"],
        exitos_err["Control"]
    ],
    nobs=[
        totales_err["Test"],
        totales_err["Control"]
    ],
    alternative="two-sided"
)

print("KPI 3 - Contraste de tasa de retrocesos")
print("\nTasas por grupo:")
print(tasa_errores)

print("\nNúmero de visitas por grupo:")
print(totales_err)

print(f"\nEstadístico Z: {z_stat:.4f}")
print(f"P-valor: {p:.6g}")
print(conclusion(p))
```

El contraste de proporciones obtiene un estadístico Z de **20,2853** y un p-valor de **1,73 × 10⁻⁹¹**, inferior al nivel de significación establecido (α = 0,05). Por tanto, rechazamos la hipótesis nula de igualdad de proporciones.

Estos resultados muestran que el nuevo diseño está asociado a una mayor frecuencia de retrocesos durante la navegación. Esto podría indicar dificultades de usabilidad que convendría investigar, especialmente en los pasos donde se producen más retrocesos.

#### 1.2. ¿Se dividieron los clientes de forma aleatoria y equitativa?

**Objetivo:** Analizar la distribución de los 50.500 clientes entre Test y Control para evaluar si los tamaños de ambos grupos están equilibrados.

Se calculará el número de clientes y su porcentaje de participación en cada grupo.

Posteriormente, se valorará si la distribución observada es compatible con una asignación equitativa y qué información sería necesaria para confirmar la aleatorización.

```python
# Distribución de clientes entre Test y Control

distribucion = clientes_ab["Variation"].value_counts()

porcentajes = (
    clientes_ab["Variation"]
    .value_counts(normalize=True) * 100
)

print("Clientes por grupo:")
print(distribucion)

print("\nPorcentaje por grupo:")
print(porcentajes.round(2))

print("\nDiferencia respecto al reparto 50/50:")
print((porcentajes - 50).round(2))
```

**Análisis de la distribución entre Test y Control**

El experimento cuenta con un total de **50.500 clientes**, distribuidos de la siguiente manera:

- Test: 26.968 clientes (53,40 %).
- Control: 23.532 clientes (46,60 %).

La distribución presenta una diferencia de 6,80 puntos porcentuales entre ambos grupos, con una mayor representación del grupo Test.

Para evaluar si esta diferencia es compatible con una asignación aleatoria equiprobable, se realizará un test binomial bilateral, suponiendo que el reparto previsto era del 50 % para cada grupo.

**Hipótesis nula (H₀):** La probabilidad de asignación al grupo Test es del 50 %.

**Hipótesis alternativa (H₁):** La probabilidad de asignación al grupo Test es diferente del 50 %.

**Nivel de significación:** α = 0,05.

Este contraste permite evaluar el equilibrio respecto a una distribución 50/50, pero no demuestra por sí solo si el procedimiento de asignación fue realmente aleatorio.

```python
from scipy.stats import binomtest

# Número de clientes asignados a Test
n_test = 26968

# Número total de clientes
n_total = 50500

# Test binomial bilateral (reparto esperado 50/50)
resultado_binomial = binomtest(
    k=n_test,
    n=n_total,
    p=0.5,
    alternative="two-sided"
)

print(f"Valor p: {resultado_binomial.pvalue:.6e}")
print(f"Proporción observada en Test: {n_test/n_total:.4f}")
```

### Resultados y conclusión: distribución de los grupos experimentales

Se ha realizado un test binomial bilateral para comprobar si la distribución de clientes entre Test y Control es compatible con una asignación aleatoria independiente con una probabilidad del 50 % para cada grupo.

**Resultados:**

- Total de clientes: 50.500.
- Grupo Test: 26.968 clientes (53,40 %).
- Grupo Control: 23.532 clientes (46,60 %).
- Valor p: 8,72 × 10⁻⁵³.
- Nivel de significación: α = 0,05.

Dado que el valor p es inferior a 0,05, **rechazamos la hipótesis nula de una distribución equiprobable 50/50**.

Esto indica que el desequilibrio observado es estadísticamente significativo y no resulta compatible con el modelo de asignación 50/50 utilizado como referencia.

**Conclusión:** Los grupos no presentan una distribución exactamente equilibrada y existe evidencia estadística de una desviación respecto al reparto 50/50.

Sin embargo, este resultado no permite afirmar que la asignación no fuera aleatoria, ya que el experimento podría haber utilizado una proporción diferente de forma intencionada.

Para confirmar si existió algún problema en la asignación, sería necesario conocer el procedimiento de aleatorización y las proporciones previstas originalmente por Vanguard.

#### 1.3. ¿Existía algún sesgo?

```python
grupos = clientes_ab.groupby("Variation")

# 1. Variables numéricas
tabla_equilibrio = pd.DataFrame({
    "Edad media": grupos["clnt_age"].mean(),
    "Edad mediana": grupos["clnt_age"].median(),
    "Saldo mediano": grupos["bal"].median(),
    "Antigüedad media (años)": grupos["clnt_tenure_yr"].mean(),
    "Antigüedad mediana (años)": grupos["clnt_tenure_yr"].median(),
    "% con 2 cuentas": grupos["num_accts"].apply(lambda x: (x == 2).mean() * 100),
})

# 2. Género en porcentaje
genero = pd.crosstab(clientes_ab["Variation"], clientes_ab["genero_grupo"],
                     normalize="index").mul(100)
genero.columns = ["% " + col for col in genero.columns]

# 3. Unir todo, girar la tabla y añadir la diferencia
tabla_equilibrio = pd.concat([tabla_equilibrio, genero], axis=1).T.round(2)
tabla_equilibrio["Diferencia (Test - Control)"] = (
    tabla_equilibrio["Test"] - tabla_equilibrio["Control"]
).round(2)

tabla_equilibrio["Diferencia relativa (%)"] = (
    (tabla_equilibrio["Test"] - tabla_equilibrio["Control"])
    / tabla_equilibrio["Control"] * 100
).round(2)

tabla_equilibrio
```

La tabla muestra que Test y Control tienen perfiles prácticamente idénticos: todas las diferencias relativas son inferiores al 2,1 %. La mayor diferencia está en la edad mediana (47,5 frente a 48,5 años, un -2,06 %), mientras que el saldo, la antigüedad, el número de cuentas y el género difieren menos de un 2 %, y la antigüedad mediana coincide exactamente. Junto con el test chi-cuadrado de género (p = 0,388), esto indica que no hay sesgos en el perfil de los clientes y que las diferencias en los KPIs pueden atribuirse al nuevo diseño.

Algo que conviene revisar: el tamaño de los grupos. Test tiene 26.968 clientes y Control 23.532, es decir, un 53,4 % frente a un 46,6 %. Si el experimento se diseñó para repartir 50/50, esa diferencia puede indicar un problema en la asignación. Y se comprueba con un chi-cuadrado:

```python
from scipy.stats import chisquare

# H0: la diferencia de tamaño entre Test y Control se debe al azar (asignación 50/50)
# H1: la diferencia de tamaño es demasiado grande para deberse al azar

observados = [26968, 23532]         
esperados = [sum(observados) / 2] * 2  

chi2_srm, p_srm = chisquare(observados, f_exp=esperados)
print(f"SRM → chi2 = {chi2_srm:.2f}, p-valor = {p_srm:.2e}")
```

Con un p-valor prácticamente igual a 0 (8,92 × 10⁻⁵³), rechazamos H₀: la diferencia de tamaño entre Test (53,4 %) y Control (46,6 %) es demasiado grande para deberse al azar en una asignación aleatoria al 50 %. Aunque ambos grupos tienen perfiles equivalentes, este desequilibrio es un posible sesgo, y convendría confirmar cuál era el reparto previsto.

Otros sesgos importantes de mencionar:

-Efecto novedad: los clientes de Test podrían comportarse distinto solo porque el diseño es nuevo, y ese efecto puede desaparecer con el tiempo.

-Efecto de costumbre: los clientes de larga trayectoria están acostumbrados al diseño antiguo, lo que podría explicar parte de los retrocesos en Test.

Sí: más de 50.000 clientes y 69.000 visitas, y todos los tests salieron muy significativos.

Si Test gana a Control todas las semanas, la conclusión es sólida. Si la ventaja solo aparece al principio y luego desaparece, sería un efecto novedad.

```python
# Rango real de fechas en los datos
print("Desde:", web_ab["date_time"].min(), "| Hasta:", web_ab["date_time"].max())

# Tasa de finalización por semana y grupo
web_ab["semana"] = web_ab["date_time"].dt.to_period("W").dt.start_time

visitas = web_ab.groupby(["Variation", "visit_id"]).agg(
    semana=("semana", "min"),
    completo=("process_step", lambda pasos: (pasos == "confirm").any())
).reset_index()

semanal = visitas.groupby(["semana", "Variation"])["completo"].mean().mul(100).unstack()

semanal.plot(marker="o")
plt.title("% de visitas que completan el proceso, por semana")
plt.xlabel("Semana")
plt.ylabel("% de visitas completadas")
plt.legend(title="Grupo")
plt.grid(axis="y", alpha=0.3)
plt.show()
```

Lo que muestra

1. Las fechas coinciden exactamente con el experimento (del 15/03/2017 al 20/06/2017), así que no hay datos fuera del periodo.
2. Hay dos etapas muy distintas:
    - De marzo a finales de abril: Test está claramente por encima de Control, a menudo con 10 o más puntos de diferencia. Control además es muy inestable y llega a caer hasta el 39 % a principios de mayo.
    - Desde mayo hasta el final: las dos líneas se juntan en torno al 55 %. Test solo está ligeramente por encima, y algunas semanas incluso por debajo (mediados de mayo y principios de junio).
3. Test baja con el tiempo: pasa de valores del 60–65 % al principio a un 55–57 % al final.

El experimento cubrió todo el periodo previsto (del 15/03 al 20/06/2017) y reunió más de 50.000 clientes, suficientes para obtener resultados estadísticamente significativos. Sin embargo, el análisis semanal muestra que la ventaja de Test no es estable: durante marzo y abril supera claramente a Control, pero desde mayo ambos grupos se igualan en torno al 55 % de visitas completadas. Esto sugiere un posible efecto novedad, por lo que la mejora real a largo plazo podría ser menor que la observada, y convendría alargar el experimento para confirmar si la diferencia se mantiene.
